Define the space and calculate the probabilities for the variable that measures the probability of getting tails when flipping two coins.

```python
print("Prob. of getting two tails: ", binom.pmf(k = 2, n = 2, p = 0.5)) 
#k = sucess, n = trials, p = probability
```

```python
print("Prob. of getting one tail: ", binom.pmf(k = 1, n = 2, p = 0.5)) 
```

```python
print("Prob. of getting zero tail: ", binom.pmf(k = 0, n = 2, p = 0.5)) 
```
